# 全微分について理解する

## 微分の定義と、微小値の存在

$ \frac{df(x)}{dx} \equiv lim_{\Delta -> 0} \frac{f(x + \Delta) - f(x)}{\Delta} $

## 変数変換のチェーンルール

$ \frac{df(s(x))}{d(x)} = \frac{df(s)}{ds} \frac{ds(x)}{dx} $

## 偏微分

$ \frac{\partial f(x,y)}{\partial x} \equiv lim_{\Delta -> 0} \frac{f(x + \Delta, y) - f(x,y)}{\Delta} $

$ \frac{\partial f(x,y)}{\partial y} \equiv lim_{\Delta -> 0} \frac{f(x, y + \Delta) - f(x,y)}{\Delta} $
